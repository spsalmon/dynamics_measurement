```python
experiments = [
    "/mnt/towbin.data/shared/spsalmon/20251014_150718_923_ZIVA_60x_443_additional_stardist_training_data/part1",
    "/mnt/towbin.data/shared/spsalmon/20251014_150718_923_ZIVA_60x_443_additional_stardist_training_data/part2",
    "/mnt/towbin.data/shared/spsalmon/20250929_161842_172_ZIVA_60x_443_stardist_training"
]

output_dir = "/mnt/towbin.data/shared/spsalmon/towbinlab_segmentation_database/stardist/emr1_panoptic_dataset_60x_auto_seg"
os.makedirs(output_dir, exist_ok=True)
```

```python

zstack_dir = "raw_zstack"

filemaps = []
zstacks = []

for experiment in experiments:
    filemap_path = os.path.join(experiment, "analysis", "report", "analysis_filemap_annotated.csv")
    if not os.path.exists(filemap_path):
        filemap_path = os.path.join(experiment, "analysis", "report", "analysis_filemap.csv")

    filemap = pd.read_csv(filemap_path)
    dir_to_add = os.path.join(experiment, zstack_dir)
    filemap = add_dir_to_experiment_filemap(filemap, dir_to_add, "raw_zstack")

    if 'Ignore' not in filemap.columns:
        filemap['Ignore'] = False

    if 'HatchTime' in filemap.columns:
        filemap['Time'] = pd.to_numeric(filemap['Time'], errors='coerce')
        # convert HatchTime to numeric
        filemap['HatchTime'] = pd.to_numeric(filemap['HatchTime'], errors='coerce')

        filemap = filemap[(filemap['HatchTime'].isna()) | (filemap['Time'] > filemap['HatchTime'])]

    filemap = filemap[~filemap['Ignore']]
    filemaps.append(filemap)
    zstack_paths = filemap['raw_zstack'].tolist()
    zstack_paths = [path for path in zstack_paths if path != ""]
    zstacks.extend(zstack_paths)
```

```python
db_size = 500
np.random.seed(42)
selected_indices = np.random.choice(len(zstacks), size=db_size, replace=False)
selected_zstacks = [zstacks[i] for i in selected_indices]

def process_zstack(zstack_path, output_dir):
    zstack_name = os.path.basename(zstack_path)
    output_path = os.path.join(output_dir, "raw", zstack_name)
    if not os.path.exists(output_path):
        shutil.copy(zstack_path, output_path)

Parallel(n_jobs=8)(delayed(process_zstack)(zstack_path, output_dir) for zstack_path in selected_zstacks)
```
